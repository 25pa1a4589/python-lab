```python
# print odd numbers upto n numbers
n=int(input("enter an numbers"))
for i in range(1,n+1,2):
  print(i,end="")
```

```python
# print multiplication of a given number
n=int(input("enter a number"))
for i in range(1,11):
  print(f"{n} *{i} ={n*i}")
```

```python
# print number is prime number or not
n=int(input("enter a number"))
for i in range(2,n):
  if(n%2==0):
    print("it is not prime number")
    break
else:
  print("it is prime number")
```

```python
# print prime numbers between 2 numbers
a=int(input("enter a number"))
b=int(input("enter a number"))
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break

  else:
    print(i,end=" ")
```

```python

# sum of digits of a given number

n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
  print(sum)
```
