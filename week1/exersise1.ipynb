```python
##largest number among three numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b and a > c:
    print("Largest number is", a)
elif b > c:
    print("Largest number is", b)
else:
    print("Largest number is", c)
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

a = a + b
b = a - b
a = a - b

print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
a = 10
b = 3

# Arithmetic operators
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

# Relational operators
print(a > b)
print(a < b)
print(a == b)

# Assignment operator
a += 2
print("After assignment:", a)

# Logical operators
print(a > 5 and b < 5)
print(a > 5 or b > 5)
print(not(a > b))

# Bitwise operators
print(a & b)
print(a | b)
print(a ^ b)

# Ternary operator
print("Even" if a % 2 == 0 else "Odd")

# Membership operator
numbers = [1, 2, 3]
print(2 in numbers)

# Identity operator
x = [1, 2]
y = x
print(x is y)
```

```python
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

for num in range(start, end + 1):
    if num > 1:
        count = 0
        for i in range(2, num):
            if num % i == 0:
                count = count + 1
        if count == 0:
            print(num)

```

```python
## match case
number = int(input("Enter a number (1 to 3): "))

match number:
    case 1:
        print("One")
    case 2:
        print("Two")
    case 3:
        print("Three")
    case _:
        print("Invalid number")

```

```python
number = int(input("Enter a number: "))
total = 0

while number > 0:
    digit = number % 10
    total = total + digit
    number = number // 10

print("Sum of digits:", total)
```
