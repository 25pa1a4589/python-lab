          TUPLE

```python
# creating tuple from users
t1= tuple(map(int,input().split()))
```

```python
print(t1)
```

```python
t2=(23,45,13,16,19)
print(t1+t2)
```

```python
#operations on tuple
print(t1+t2) #concatination
print(t1*2) #repeatation
print(t1[3],t2[3]) #indexing
print(t1[-1]) #negative indexing
print(t1[1:4]) #slicing
print(len(t2)) #length
```

```python
#functions on tuples
print(sum(t1+t2))
print(max(t2))
print(max(t1))
print(len(t1))
print(len(t2))
print(min(t2))
print(sorted(t1))
```

```python
#methods on tuples
t=(10,45,28,10,56)
print(t.count(10)) #count method
print(t.index(10))  #indexing method
```

```python
#zip
print(tuple((zip(t1,t2))))
```

```python
t3=(2,5,67,34,[56,4,89])
print(t3[3][2])
```

```python
#print two students information
s1=("siva",19,"bhi","vit")
s2=("prabhas",20,"mgt","vit")
print(s1+s2)
```

```python
#counting no.of vowels
a=str(input("enter a word"))
b=a.lower()
vowel=b.count('a')+b.count("e")+b.count("i")+b.count("o")+b.count("u")
print(vowel)
```
